```python
import pandas as pd
import numpy as np
import math

# First column: odd numbers between 1 and 25 inclusive
a = list(range(1, 26, 2))

# Second column: (25 - a) / 2 to find the no. of 2s that'd correspond to first collumns no. of 1s so that they sum to 25
b = [(25 - x) // 2 for x in a]

# Third column: sum of the first two columns
l = [x + y for x, y in zip(a, b)]


table = pd.DataFrame({
    'a': a,
    'b': b,
    'l': l,
    '(1/2)^l': [0.5 ** x for x in l],
    'l! / (a!b!)': [ math.factorial(z) // (math.factorial(x) * math.factorial(y)) for x, y, z in zip(a, b, l) ],
    'Product': [ (0.5 ** z) * math.factorial(z) / (math.factorial(x) * math.factorial(y)) for x, y, z in zip(a, b, l) ] })

print(table.to_string(index=False))
print("Sum of last column:", table['Product'].sum())
```
